```python
import pandas as pd
from sklearn import svm, datasets
```

```python
df = pd.read_csv("/content/Iris.csv")
```

```python
df.head()
```

```python
df.columns
```

```python

X = df.drop(columns=['Id','Species'])
y = df['Species']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,y, test_size=0.3)
```

```python
model = svm.SVC(kernel='rbf',C=30,gamma='auto')
model.fit(x_train,y_train)
model.score(x_test,y_test)
```

**K-fold Cross Validation**

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(svm.SVC(kernel = 'linear', C=10, gamma='auto'),X,y,cv=5)
```

```python
cross_val_score(svm.SVC(kernel = 'linear', C=20, gamma='auto'),X,y,cv=5)
```

```python
cross_val_score(svm.SVC(kernel = 'linear', C=30, gamma='auto'),X,y,cv=5)
```

```python
import numpy as np
kernels = ['rbf','linear']
C = [10,20,30]
avg_scores = {}
for kval in kernels:
    for cval in C:
        cv_scores = cross_val_score(svm.SVC(kernel=kval, C=cval, gamma='auto'), X, y, cv=5)
        avg_scores[kval + '-' + str(cval)] = np.average(cv_scores)

avg_scores

```

```python
from sklearn.model_selection import GridSearchCV

clf = GridSearchCV(svm.SVC(gamma = 'auto'), {
    'C' : [10,20,30],
    'kernel' : ['rbf','linear']

}, cv=5, return_train_score=False)


clf.fit(x_train,y_train)
clf.cv_results_
```

```python
df = pd.DataFrame(clf.cv_results_)
df
```

```python
df[['param_C','param_kernel','mean_test_score']]
```
